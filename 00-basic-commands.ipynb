```python
file_base_path = 'dbfs:/Volumes/workspace/default/udemy-files-class/arquivos_csv'
clients_file_path = f'{file_base_path}/Clientes.csv'

df_clients = spark.read.format('csv') \
    .option('header', 'true') \
    .option('inferSchema', 'true') \
    .option('delimiter', ',') \
    .load(clients_file_path)

display(df_clients)


```

# Transform the Dataframe as table

```python
# it will save in the default catalog
df_clients.write.mode('overwrite').saveAsTable('clients')
```
